# To connect to your data you will need to go through these steps:

## Manual Steps:

1. Connect to a VPN, so that you appear to be 'within the university library'. You do this by hand, usually using the FortiClient app. See tech support pages (https://information-services.ed.ac.uk/computing/desktop-personal/vpn) on how to set it up (if you haven't already)

## R code steps:

2. Connect to your secure 'online harddrive' (called DataStore) using `connectDatastore(...)`, see examples. Below code will create a 'virtual connection' to your DataStore harddrive / folder. By default, this is set to be named `R:` (on windows) or `~/.temporary-data` (on a mac). In an unlikely event of you already having a drive/folder of this name, you'll need to tweak the code.
3. You can peek into the DataStore folder with `openDatastoreFolder()` - it will open file browser, and appear as if it was a new harddrive or folder on your computer. Do not copy files from there by hand. Use R to load that data instead.
4. Now you can use the DataStore files. Load and save the data using provided functions like `loadCsvWithDatastore(your_filename)` or `saveCsvWithDatastore(your_filename, some_data)`. Advanced: There are various helper functions, but if you need to build your own, e.g. you use unusual data format, just use `fileLocationOnDatastore(your_filename)` instead of `your_filename` to get the file's location for saving and loading. So e.g. instead of `load_csv("patients.tsv")` use `load_csv( fileLocationOnDatastore("patients.tsv"))`
5. When done with accessing data, you can disconnect from DataStore with `disconnectDatastore()`
